```python
# kb = [

#    ({"engine_not_starting", "clicking"}, "weak_battery"),

#    ({"weak_battery"}, "charge_battery"),

#    ({"engine_hot", "low_coolant"}, "coolant_problem"),

#    ({"coolant_problem"}, "add_coolant"),

#    ({"brake_noise"}, "brake_problem"),

#    ({"brake_problem"}, "check_brakes"),

#    ({"fuel_low"}, "fuel_problem"),

#    ({"fuel_problem"}, "add_fuel")

# ]

 

facts = set()

rules = [
    ({"yellow_leaves", "white_powder"}, "powdery_mildew"),
    ({"brown_roots", "wilting"}, "root_rot"),
    ({"brown_spots", "yellow_leaves"}, "leaf_spot"),
    ({"black_spots", "wilting"}, "fungal_disease")
]


# Forward Chaining
def forward_chaining():

    new_fact = True

    while new_fact:
        new_fact = False

        for conditions, result in rules:
            if conditions.issubset(facts) and result not in facts:
                facts.add(result)
                print("Derived:", result)
                new_fact = True


# Backward Chaining
def backward_chaining(goal, checked=None):

    if checked is None:
        checked = set()

    if goal in facts:
        return True

    if goal in checked:
        return False

    checked.add(goal)

    for conditions, result in rules:

        if result == goal:

            possible = True

            for condition in conditions:
                if not backward_chaining(condition, checked):
                    possible = False
                    break

            if possible:
                return True

    return False


# Main Program
while True:

    print("\n-------- MENU --------")
    print("1. Forward Chaining")
    print("2. Backward Chaining")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1" or choice == "2":

        facts.clear()
        symptoms = input("Enter symptoms separated by comma: ").lower().split(",")
        for symptom in symptoms:
            facts.add(symptom.strip())

        
        if choice == "1":
            forward_chaining()

            print("Final Facts:", facts)
            diseases = [ "powdery_mildew", "root_rot","leaf_spot","fungal_disease"]

            for disease in diseases:
                if disease in facts:
                    print("Disease detected:", disease)

        else:

            goal = input("Enter disease to check: ").lower().strip()
            
            if backward_chaining(goal):
                print("Disease confirmed:", goal)
            else:
                print("Disease cannot be confirmed.")

    elif choice == "3":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
